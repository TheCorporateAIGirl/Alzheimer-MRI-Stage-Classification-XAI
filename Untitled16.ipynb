```python
import zipfile
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models
from sklearn.metrics import confusion_matrix, classification_report

# 1. Extraction
print("📦 Extracting data...")
with zipfile.ZipFile('archive.zip', 'r') as zip_ref:
    zip_ref.extractall('./final_sprint')

# 2. SMART Path Finding
# This looks for any folder containing "MildDemented"
train_dir = ""
for root, dirs, files in os.walk('./final_sprint'):
    if 'MildDemented' in dirs:
        train_dir = root
        break

if not train_dir:
    # If it still can't find it, let's list everything to see what's inside
    print("❌ Could not find classes. Available folders:")
    print(os.listdir('./final_sprint'))
else:
    print(f"✅ Found data at: {train_dir}")
    print(f"Stages detected: {os.listdir(train_dir)}")

    # 3. Data Generators
    print("🖼️ Preparing images...")
    datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

    train_gen = datagen.flow_from_directory(
        train_dir,
        target_size=(176, 176),
        batch_size=32,
        subset='training'
    )

    val_gen = datagen.flow_from_directory(
        train_dir,
        target_size=(176, 176),
        batch_size=32,
        subset='validation'
    )

    # 4. Build Model
    print("🏗️ Building MobileNetV2...")
    base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(176, 176, 3))
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(4, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # 5. Fast Training Sprint
    print("🚀 Training starting... (Will take ~10 mins)")
    model.fit(train_gen, steps_per_epoch=20, epochs=3, validation_data=val_gen, validation_steps=10)

    # 6. Final Results
    print("📊 Generating Clinical Report...")
    val_gen.reset()
    x, y_true_cat = next(val_gen)
    y_pred = np.argmax(model.predict(x), axis=1)
    y_true = np.argmax(y_true_cat, axis=1)

    plt.figure(figsize=(8, 6))
    sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, cmap='Blues',
                xticklabels=list(val_gen.class_indices.keys()),
                yticklabels=list(val_gen.class_indices.keys()))
    plt.title('Final Result: Alzheimer Classification')
    plt.show()
```

```python
import tensorflow as tf
import numpy as np
import cv2

def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    # 1. Create a model that maps the input image to the activations of the last conv layer
    # This acts as the feature extractor part
    feature_extractor = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=model.get_layer('mobilenetv2_1.00_224').get_layer(last_conv_layer_name).output
    )

    # 2. Create a model that maps the activations of the last conv layer to the final predictions
    # This acts as the classifier part, skipping the base_model layer
    classifier_inputs = tf.keras.Input(shape=feature_extractor.output.shape[1:])
    x = classifier_inputs
    for layer in model.layers[1:]: # Iterate through the layers after the base_model
        x = layer(x)
    classifier_model = tf.keras.models.Model(inputs=classifier_inputs, outputs=x)

    # Compute the gradient of the top predicted class for our input image
    with tf.GradientTape() as tape:
        # Get the activations of the last conv layer
        last_conv_layer_output = feature_extractor(img_array)
        tape.watch(last_conv_layer_output)
        # Get the predictions from the classifier part
        preds = classifier_model(last_conv_layer_output)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    # This is the gradient of the output neuron with regard to the output feature map
    grads = tape.gradient(class_channel, last_conv_layer_output)

    # This is a vector where each entry is the mean intensity of the gradient over a specific feature map channel
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    # We multiply each channel in the feature map array by "how important this channel is"
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    # Normalize the heatmap between 0 & 1 for visualization
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()
```

```python
def display_gradcam(img_path, model, last_conv_layer_name):
    img = tf.keras.preprocessing.image.load_img(img_path, target_size=(176, 176))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    # Generate heatmap
    heatmap = make_gradcam_heatmap(img_array, model, last_conv_layer_name)

    # Rescale heatmap to a range 0-255
    heatmap = np.uint8(255 * heatmap)
    jet = plt.cm.get_cmap("jet")

    # Use RGB values of the colormap
    jet_colors = jet(np.arange(256))[:, :3]
    jet_heatmap = jet_colors[heatmap]

    # Create an image with RGB colorized heatmap
    jet_heatmap = tf.keras.preprocessing.image.array_to_img(jet_heatmap)
    jet_heatmap = jet_heatmap.resize((176, 176))
    jet_heatmap = tf.keras.preprocessing.image.img_to_array(jet_heatmap)

    # Superimpose the heatmap on original image
    superimposed_img = jet_heatmap * 0.4 + (img_array[0] * 255)
    superimposed_img = tf.keras.preprocessing.image.array_to_img(superimposed_img)

    # Display
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title("Original MRI")
    plt.subplot(1, 2, 2)
    plt.imshow(superimposed_img)
    plt.title("AI Focus Area (Grad-CAM)")
    plt.show()

# Find the name of the last conv layer in MobileNetV2
# It is usually 'Conv_1' or 'out_relu'
last_layer = "out_relu"

# Test it on a random image from your validation set
sample_img = os.path.join(train_dir, 'ModerateDemented', os.listdir(os.path.join(train_dir, 'ModerateDemented'))[0])
display_gradcam(sample_img, model, last_layer)
```

```python
import zipfile
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix

# 1. Extraction
if os.path.exists('archive.zip'):
    with zipfile.ZipFile('archive.zip', 'r') as zip_ref:
        zip_ref.extractall('./final_data')

    # 2. Path Finding
    train_dir = ""
    for root, dirs, files in os.walk('./final_data'):
        if 'train' in dirs:
            train_dir = os.path.join(root, 'train')
            break

    # 3. Data Generators
    datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)
    train_gen = datagen.flow_from_directory(train_dir, target_size=(176, 176), subset='training')
    val_gen = datagen.flow_from_directory(train_dir, target_size=(176, 176), subset='validation')

    # 4. Model Setup (Fast MobileNetV2)
    base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(176, 176, 3))
    base_model.trainable = False
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dense(4, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # 5. Sprint Training
    print("Starting fast training...")
    model.fit(train_gen, steps_per_epoch=20, epochs=3)

    # 6. Final Confusion Matrix
    val_gen.reset()
    x, y_true_cat = next(val_gen)
    y_pred = np.argmax(model.predict(x), axis=1)
    y_true = np.argmax(y_true_cat, axis=1)

    plt.figure(figsize=(8, 6))
    sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, cmap='Blues',
                xticklabels=val_gen.class_indices.keys(),
                yticklabels=val_gen.class_indices.keys())
    plt.title('Final Result: Alzheimer Classification')
    plt.show()
else:
    print("❌ archive.zip not found. Please wait for the upload to finish in the sidebar.")
```

```python
import zipfile
import os
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from sklearn.metrics import confusion_matrix

# 1. Verification and Extraction
if os.path.exists('archive.zip'):
    print("✅ File found! Unzipping now...")
    with zipfile.ZipFile('archive.zip', 'r') as zip_ref:
        zip_ref.extractall('./final_data')

    # 2. Path Finding
    train_dir = ""
    for root, dirs, files in os.walk('./final_data'):
        # Looking for the folder that contains the actual classes
        if 'MildDemented' in dirs or 'NonDemented' in dirs:
            train_dir = root
            break

    if train_dir:
        print(f"📍 Data located at: {train_dir}")

        # 3. Fast Data Loading
        datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)
        train_gen = datagen.flow_from_directory(train_dir, target_size=(176, 176), batch_size=32, subset='training')
        val_gen = datagen.flow_from_directory(train_dir, target_size=(176, 176), batch_size=32, subset='validation')

        # 4. Building the Model (MobileNetV2 for speed)
        base_model = tf.keras.applications.MobileNetV2(weights='imagenet', include_top=False, input_shape=(176, 176, 3))
        base_model.trainable = False

        model = models.Sequential([
            base_model,
            layers.GlobalAveragePooling2D(),
            layers.Dense(128, activation='relu'),
            layers.Dense(4, activation='softmax')
        ])
        model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

        # 5. The Sprint Training (5-10 minutes)
        print("🚀 Starting final sprint training...")
        model.fit(train_gen, steps_per_epoch=20, epochs=3, verbose=1)

        # 6. Generate the Confusion Matrix
        print("📊 Generating Results...")
        val_gen.reset()
        x, y_true_cat = next(val_gen)
        y_pred = np.argmax(model.predict(x), axis=1)
        y_true = np.argmax(y_true_cat, axis=1)

        plt.figure(figsize=(8, 6))
        cm = confusion_matrix(y_true, y_pred)
        sns.heatmap(cm, annot=True, cmap='Blues',
                    xticklabels=list(val_gen.class_indices.keys()),
                    yticklabels=list(val_gen.class_indices.keys()))
        plt.title('Alzheimer Disease Classification: Final Result')
        plt.ylabel('Actual Stage')
        plt.xlabel('Predicted Stage')
        plt.show()
    else:
        print("❌ Found the zip, but couldn't find the 'train' folders inside. Check your zip structure.")
else:
    print("❌ archive.zip is STILL not in the sidebar. Please re-upload it.")
```

```python
import matplotlib.pyplot as plt

# Even if you stop early, you can plot the progress made so far
def plot_results(history):
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    auc = history.history['auc']
    val_auc = history.history['val_auc']
    epochs = range(1, len(acc) + 1)

    plt.figure(figsize=(12, 5))

    # Plot Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(epochs, acc, 'bo', label='Training Acc')
    plt.plot(epochs, val_acc, 'b', label='Validation Acc')
    plt.title('Training and Validation Accuracy')
    plt.legend()

    # Plot AUC
    plt.subplot(1, 2, 2)
    plt.plot(epochs, auc, 'ro', label='Training AUC')
    plt.plot(epochs, val_auc, 'r', label='Validation AUC')
    plt.title('Training and Validation AUC')
    plt.legend()

    plt.show()

# plot_results(history)
```
